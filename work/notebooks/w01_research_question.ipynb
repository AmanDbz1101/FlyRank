## 4. Careful words: what I can and can't claim

**What this work can claim:**
- **Observed** associations between content signals (age, position, CTR, engagement) and traffic decline.
- **Directional** rankings: page A shows stronger decline signals than page B.
- **Decision-support** scores that help an editor triage their queue.

**What this work will never claim:**
- Causal proof that refreshing a page *caused* a recovery — that requires an experiment, which this observational data cannot provide.
- Predicting Google's algorithm — we measure what happened, not what Google will do.
- A guarantee that pages ranked highly will recover — the output is a review priority, not a recovery forecast.
